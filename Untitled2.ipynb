```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
from sklearn.linear_model import LogisticRegression
# Separação dos dados de treino e de teste
# Usaremos 20% para teste e 80% para treino
# X_train; Y_train; X_test; Y_test
from sklearn.model_selection import train_test_split
from sklearn.metrics import accuracy_score
```

```python
dados = pd.read_csv('/electrical+grid+stability+simulated+data.zip')
```

```python
dados.head()
```

```python
dados['stabf'].value_counts()
```

```python
#Inspeção inicial
dados.shape
```

```python
dados.info()
```

```python
# Atributos comvalores numéricos
dados.describe(include='number')
```

```python
# Atributos comvalores categóricos
dados.describe(include='object')
```

```python
# Renomeando os labels
dados['stabf'] = dados['stabf'].replace ({'stabf': 'REDE ESTÁVEL', 'unstable': 'REDE INSTÁVEL'})
dados.head
```

```python
# Quais classes temos na coluna 'stabf' (target)?
dados['stabf'].unique()
```

```python
# Quais classes temos na coluna 'stabf' (target)?
dados['stabf'].value_counts()
```

```python
X = dados.drop(['stab', 'stabf'], axis=1)
X.head()
```

```python
Y = dados['stabf']
Y.head()
```

```python
# Separando dados para treino e para teste
X_train, X_test, y_train, y_test = train_test_split(X, Y, test_size=0.2, random_state=42)
```

```python
modelo = LogisticRegression()
```

```python
# Treinamento
modelo.fit(X_train, y_train)
```

```python
# Probabilidades das previsões
probabil = modelo.predict_proba(X_test)
probabil[:30]
```

```python
# Gerando o array com as previsões
y_predict = modelo.predict(X_test)
y_predict[:30]
```

```python
# Métrica de avaliação
acuracia = accuracy_score(y_test, y_predict)
print(f'Acurácia: {acuracia*100:.2f}%')
```
